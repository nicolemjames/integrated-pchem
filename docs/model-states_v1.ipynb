# Modeling States

Theoretical and mathematical models can be used to describe and predict properties and behavior of systems. These models often have conceptual implications. If the model matches experimental data, unpacking the "story" the model conveys can deepen our understanding of _how_ and _why_ systems behave the way they do. This chapters explores how to develop, use, and evaluate theoretical and mathematical models for the state of a gaseous system. 

**Why states?** The state of a system is static: a snapshot in time. This is useful in and of itself, but it's also a first step in being able to describe system _dynamics_. 

**Why gases?** Gases are a common[^gases] form of matter, and they are useful to be able to describe. They are also a nice case study for modelling states in general, because the characteristics we need to model are simpler (or at least more straightforward) than those for most non-gaseous systems.

[^gases]: I suspect you are surrounded by them right now.

## Equations of State


The state of the system is defined by its physical properties. In order to develop an equation of state for a system, we need to determine what variables influence its properties, and how the properties depend on those variables. In other words, we want to develop a mathematical relationship that lets us represent a parameter as a function of all other relevant parameters. 

There are many ways to try to develop such a mathematical model. Broadly speaking, the two "main" traditions for this in thermodynamics are **classical** approaches and **statistical** approaches. 

Classical thermodynamics is older. In fact, it precedes the scientific consensus on the existence of atoms. Classical models are empirically derived from observations of _macroscopic_ properties of bulk matter. This has several advantages, and can relatively directly be used to determine how macroscopic properties (e.g., pressure, volume, temperature) are related. However, it has limited capacity to explain _how_ or _why_ those relationships are what they are. Essentially, it experimentally examines bulk relationships and might speculate some theories about how the particulate nature gives rise to them, but is primarily a "top-down" approach.

Statistical thermodynamics, in contrast, models system based on our theoretical understanding of the particulate component(s) and derives predictions of bulk  properties by statistically "scaling up" the behavior of individual particulates. This "bottom-up" approach is advantageous for explaining _how_ or _why_ particulate (e.g., molecular) behaviors give rise to macroscopic properties. However, a limitation is: if the model is "wrong" (and all models are), the predictions might not match reality. 

Thus, physical chemistry will often draw on both classical and statistical mechanics as needed, based on which is most relevant (or useful) to unpacking whatever it is we are trying to understand about a system. 

Models can be derived classically or statistically. One particularly compelling way to build an argument that a statistical mechanics-derived approach is "true"[^true] is to show that it agrees with empirically-derived classical predictions. The ideal gas law is a nice example of this. 

[^true]: defining what it means for something to be "true" is, it turns out, not simple--especially when discussing models, all of which are flawed in some way.

### The Ideal Gas Equation of State

Physical quantities needed to specify thestate of a gaseous system include the amount of substance $n$, the volume $V$ it occupies, its temperature $T$, and its pressure $P$. If these quantities are sufficient to describe the entire system, we should be able to formulate one in terms of the others. In other words: if $n$, $V$, $T$, and $P$ are a full minimal set of **state variables**, then we should be able to formulate a function: 

$$P(T, V, n)$$

Consider the difference between the classical and statistical approaches: 

::::{grid} 1 1 2 2
:gutter: 3

:::{grid-item-card} Classical Approach
:class-card: sd-shadow-sm

Starting point: apparatuses that enable measurement of $P$, $V$, $n$, $T$. 

1. Measure how pressure and volume are related at constant $n$ and $T$:

   $$P \propto \frac{1}{V}$$

2. Measure how pressure and temperature are related at constant $n$ and $V$:

   $$P \propto T$$

3. Measure how volume and amount of gas are related at constant $P$ and $T$:

   $$V \propto n$$

4. Combine the empirical proportionalities:

   $$P \propto \frac{nT}{V}$$

5. Introduce the proportionality constant ($R$), which can be determined by fitting the experimental data:

   $$P = \frac{nRT}{V}$$

:::

:::{grid-item-card} Statistical Approach
:class-card: sd-shadow-sm

Starting point: a model for motion of non-interactive microscopic "corpuscules"[^corp] with negliglible volume.

1. Assume that all internal energy of particles comes from translational kinetic energy:

   $$KE = \frac{1}{2}mv^2$$

2. Relate pressure to one particle's collisions with the container wall in the $x$ direction. 

3. Count the momentum transferred by all particle-wall collisions over a time interval $\Delta t$, assuming collisions with other particles are perfectly elastic.

4. Sum over all particles to obtain the pressure from molecular motion. 

5. Assume particles independently in all three directions ($x$, $y$, $z$), use the theoretical relationship between temperature and average translational kinetic energy:

   $$\langle KE \rangle = \frac{3}{2}k_BT$$

6. Convert from particles to moles using $N = nN_A$ and $R = N_Ak_B$, yielding:

      $$P = \frac{nRT}{V}$$

Watch [this video](https://youtu.be/leuIr_zDk18?si=gbUzuooC01-DSJkB) to see a complete demonstration of each of these steps.
:::

::::

[^corp]: A term commonly used for a discrete particle of matter before concepts of "atoms" and "molecules" were introduced. For example, in his 1738 book Hydrodynamica, Daniel Bernoulli wrote _"let the cavity contain very minute corpuscles, which are driven hither and thither with a very rapid motion; so that these corpuscles, when they strike against the piston and sustain it by their repeated impacts, form an elastic fluid which will expand of itself if the weight is removed or diminished..."_


To recap: both models yield the same end result--the ideal gas equation of state is $P=nRT/V$.[^ideal-gas] 

The classical approach does this by measuring and mathematically manipulating macroscopic, bulk-scale experimental data. The statistical approach does this by drawing on existing models for the particulate structure of the system, which themselves invoke several theoretical assumptions: 

1. The volume occupied by particles is neglibly different from the container volume
2. Collisions are perfectly elastic
3. Particles move independently, in a straight line until they collide with the container wall.
4. The only form of energy in the system is translational kinetic energy.
5. By the **equipartition theorem**, translational kinetic energy is distributed equally in the $x$, $y$, and $z$ directions.

[^ideal-gas]: or, equivalently: $PV = nRT$

The assumption that all energy is translational kinetic energy suggests that other energetic modes (vibrations, rotations, etc) are not active. We will discuss this more in the coming sections, for now we'll ignore this. 

The assumption that kinetic energy is distributed equally in $x$, $y$, $z$ directions suggests there are no directional forces (such as gravity). In most cases, atoms and molecules have such low mass that we will gleefully ignore gravity. 

The ideal gas equation of state should "work" best for systems where the assumptions made to develop the model hold best. In particular, the assumption that the volume occupied by particles is neglibly different from the container volume (i.e, that $V_\text{container} = V$) is only "true" if particles take up no volume themselves. Thus, the higher the volume occupied by particles, the less reasonable this assumption. The volume occupied by particles increases as the molecular size increases, and as the number density ($n/V$) increases. As a reference point, at standard temperature and pressure (0 K, 1 atm), 1 mol of an ideal gas occupies 22.4 L. 

The assumptions that particles move independently of one another, in a straight line until they collide (perfectly elastically) with the wall, will begin to break down if particles interact with one another (for example, through intermolecular forces). 

With this in mind: the ideal gas equaton of state should "work" best for the $V=1$ L system because it has the largest volume per mole (lowest number density $n/V$) of the three systems, such that the fraction of the volume in the container occupied by particles is lowest. Conversely, the ideal gas equation of state should be expected to work least well for the $V = 0.01$ L system, because it has the lowest volume per mole (highest $n/V$),  and highest volume fraction of particles.

The plots in {ref}`2D-state` show 2-dimensional slices through a higher-dimensional set of all the possible states. Since $P(n,V,T)$ would require 4 dimensions to visualize (one for each variable), we will hold one variable (commonly $n$) fixed, and visualize one variable ($P$) as a function of all possible combinations of the others ($V,T$).

Here, the plot on the left displays the state space for the 1.25 moles of an ideal gas over roughly the same range used in {ref}`2D-state`. The constant-volume slices you previously plotted are shown as lines. You can drag the figure to rotate the surface!

The figure on the right shows the same information, but for 2.5 moles. Notice that changing the amount of moles doesn't change the shape of the surface - it just scales it.

````python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

R = 0.08206 # L atm / mol K
n_1 = 1.25 # mol
n_2 = 2.5 # mol
T_values = np.linspace(79, 298, 80) # K
V_values = np.linspace(0.01, 1.0, 80) # L
curve_volumes = [1.0, 0.1, 0.01] # L

def ideal_pressure(T, V, n):
    return n * R * T / V

def plot_state_surface(ax, T_values, V_values, pressure_model, n, curve_volumes):
    T_grid, V_grid = np.meshgrid(T_values, V_values)
    P_grid = pressure_model(T_grid, V_grid, n)

    surface = ax.plot_surface(T_grid, V_grid, P_grid, cmap='viridis',
                              alpha=0.4, linewidth=0)


    colors = ['blue', 'orange', 'red']
    for V, color in zip(curve_volumes, colors):
        P_curve = pressure_model(T_values, V, n)
        V_curve = np.full_like(T_values, V)
        ax.plot(T_values, V_curve, P_curve, color=color, linewidth=3,
                label=f'V = {V:g} L')

    ax.set_xlabel('Temperature (K)')
    ax.set_ylabel('Volume (L)')
    ax.set_zlabel('Pressure (atm)')
    ax.set_yticks([0.01, 0.1, 1.0])
    ax.view_init(elev=20, azim=35)
    ax.set_title(f'n = {n:g} mol')
    ax.legend(loc='upper left')
    return surface, P_grid.max()

fig = plt.figure(figsize=(12, 5))
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
ax2 = fig.add_subplot(1, 2, 2, projection='3d')

surface1, zmax1 = plot_state_surface(ax1, T_values, V_values, ideal_pressure, n_1, curve_volumes)
surface2, zmax2 = plot_state_surface(ax2, T_values, V_values, ideal_pressure, n_2, curve_volumes)

zmax = max(zmax1, zmax2)
ax1.set_zlim(0, zmax * 1.05)
ax2.set_zlim(0, zmax * 1.05)

fig.suptitle('Ideal-gas P(V,T) state surfaces')
fig.colorbar(surface2, ax=[ax1, ax2], shrink=0.6, pad=0.08, label='Pressure (atm)')
plt.show()
````
