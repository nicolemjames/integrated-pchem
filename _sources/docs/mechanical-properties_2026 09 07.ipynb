# Mechanical Properties (P,V,n)

Theoretical and mathematical models can be used to describe and predict properties and behavior of systems. These models often have conceptual implications. If the model matches experimental data, unpacking the "story" the model conveys can deepen our understanding of _how_ and _why_ systems behave the way they do. This chapters explores how to develop, use, and evaluate theoretical and mathematical models predicting {term}`mechanical properties` (e.g., $P$,$V$, $n$) of the state of gaseous systems. 

:::{admonition} Side Note
:class: hint margin simple

**Why states?** The state of a system is static: a snapshot in time. This is useful in itself, and the first step toward describing system _dynamics_. 

**Why mechanical properties?** Mechanical properties are critical for explaining system behavior, are (often) empirically measurable, and a useful foundation for describing non-mechanical properties.

**Why gases?** Gases are a common[^gases] form of matter. They are also a nice case study because the behaviors we need to model are more straightforward than most in non-gaseous systems.
:::

In order to develop an {term}`equation of state` for a system, we need to determine what variables influence its properties, and how the properties depend on those variables. In other words, we want to develop a mathematical relationship that lets us represent a parameter as a function of all other relevant parameters. There are many ways to try to develop such a mathematical model. Broadly speaking, the two "main" traditions for this in thermodynamics are **classical**  and **statistical**. 

Classical thermodynamics is older. In fact, it precedes the scientific consensus on the existence of atoms. Classical models are empirically derived from observations of _macroscopic_ properties of bulk matter. This has several advantages, and can relatively directly be used to determine how macroscopic properties (e.g., pressure, volume, temperature) are related. However, it has limited capacity to explain _how_ or _why_ those relationships are what they are. Essentially, it experimentally examines bulk relationships and might speculate some theories about how the particulate nature gives rise to them, but is primarily a "top-down" approach.

Statistical thermodynamics, in contrast, models system based on our theoretical understanding of the particulate component(s) and derives predictions of bulk  properties by statistically "scaling up" the behavior of individual particulates. This "bottom-up" approach is advantageous for explaining _how_ or _why_ particulate (e.g., molecular) behaviors give rise to macroscopic properties. However, a limitation is: if the model is "wrong" (and all models are), the predictions might not match reality. One particularly compelling way to build an argument that a statistical mechanics-derived approach is "true"[^true] is to show that it agrees with empirically-derived classical predictions.

[^true]: defining what it means for something to be "true" is, it turns out, not simple--especially when discussing models, all of which are flawed in some way. 

[^gases]: I suspect you are surrounded by them right now.

## The Ideal Gas Equation of State

Physical quantities needed to specify thestate of a gaseous system include the amount of substance $n$, the volume $V$ it occupies, its temperature $T$, and its pressure $P$. If these quantities are sufficient to describe the entire system, we should be able to formulate one in terms of the others. In other words: if $n$, $V$, $T$, and $P$ are a full minimal set of **state variables**, then we should be able to formulate a function: 

$$P(T, V, n)$$

Consider the difference between the classical and statistical approaches:

:::{grid-item-card} Classical Approach
:class-card: sd-shadow-sm

Starting point: apparatuses that enable measurement of $P$, $V$, $n$, $T$. 

1. Measure how pressure and volume are related at constant $n$ and $T$:

   $$P \propto \frac{1}{V}$$

2. Measure how pressure and temperature are related at constant $n$ and $V$:

   $$P \propto T$$

3. Measure how volume and amount of gas are related at constant $P$ and $T$:

   $$V \propto n$$

4. Combine the empirical proportionalities:

   $$P \propto \frac{nT}{V}$$

5. Introduce the proportionality constant ($R$), which can be determined by fitting the experimental data:

   $$P = \frac{nRT}{V}$$

:::

:::{grid-item-card} Statistical Approach
:class-card: sd-shadow-sm

Starting point: a model for motion of non-interactive microscopic "corpuscules"[^corp] with negliglible volume.

1. Assume that all internal energy of particles comes from translational kinetic energy:

   $$KE = \frac{1}{2}mv^2$$

2. Relate pressure to one particle's collisions with the container wall in the $x$ direction. 

3. Count the momentum transferred by all particle-wall collisions over a time interval $\Delta t$, assuming collisions with other particles are perfectly elastic.

4. Sum over all particles to obtain the pressure from molecular motion. 

5. Assume particles independently in all three directions ($x$, $y$, $z$), use the theoretical relationship between temperature and average translational kinetic energy:

   $$\langle KE \rangle = \frac{3}{2}k_BT$$

6. Convert from particles to moles using $N = nN_A$ and $R = N_Ak_B$, yielding:

      $$P = \frac{nRT}{V}$$

Watch [this video](https://youtu.be/leuIr_zDk18?si=gbUzuooC01-DSJkB) to see a complete demonstration of each of these steps.
:::

::::

[^corp]: A term commonly used for a discrete particle of matter before concepts of "atoms" and "molecules" were introduced. For example, in his 1738 book Hydrodynamica, Daniel Bernoulli wrote _"let the cavity contain very minute corpuscles, which are driven hither and thither with a very rapid motion; so that these corpuscles, when they strike against the piston and sustain it by their repeated impacts, form an elastic fluid which will expand of itself if the weight is removed or diminished..."_


To recap: both models yield the same end result--the ideal gas equation of state is $P=nRT/V$.[^ideal-gas] 

The classical approach does this by measuring and mathematically manipulating macroscopic, bulk-scale experimental data. The statistical approach does this by drawing on existing models for the particulate structure of the system, which themselves invoke several theoretical assumptions: 

1. The volume occupied by particles is neglibly different from the container volume
2. Collisions are perfectly elastic
3. Particles move independently, in a straight line until they collide with the container wall.
4. The only form of energy in the system is translational kinetic energy.
5. Translational kinetic energy is distributed equally in the $x$, $y$, and $z$ directions.[^equipart]

[^ideal-gas]: or, equivalently: $PV = nRT$
[^equipart]: per the equipartition theorem

The assumption that all energy is translational kinetic energy suggests that other energetic modes (vibrations, rotations, etc) are not active. We will discuss this more in the coming sections--for helium, this is a valid assumption.  

The assumption that kinetic energy is distributed equally in $x$, $y$, $z$ directions suggests there are no directional forces (such as gravity). In most cases, atoms and molecules have such low mass that we will gleefully ignore gravity. There are no other directional forces afoot in this system. 

The assumption that the volume occupied by particles is neglibly different from the container volume (i.e, that $V_\text{container} = V$) is only "true" if particles take up no volume themselves. Thus, the higher the volume occupied by particles, the less reasonable this assumption. The volume occupied by particles increases as the molecular size increases, and as the number density ($n/V$) increases. As a reference point, at standard temperature and pressure (0 K, 1 atm), 1 mol of an ideal gas occupies 22.4 L. Helium is the smallest neutral atom, so this assumption should work better for this system than it it does for literally any other atomic or molecular system.

The assumptions that particles move independently of one another, in a straight line until they collide (perfectly elastically) with the wall, will begin to break down if particles interact with one another (for example, through intermolecular forces). Helium is the least polarizable element, so again this assumption should work better for helium than for any other system. 

With this in mind: the ideal gas equaton of state should "work" reasonably well for this system. The main consideration is the density of particles: this should work best for the $V=1$ L system because it has the largest volume per mole (lowest number density $n/V$) of the three systems, such that the fraction of the volume in the container occupied by particles is lowest. Conversely, the ideal gas model should work least well for the $V = 0.01$ L system, because it has the lowest volume per mole (highest $n/V$),  and highest volume fraction of particles.

The plots in {ref}`2D-state` show 2-dimensional slices through a higher-dimensional set of all the possible states. Since $P(n,V,T)$ would require 4 dimensions to visualize (one for each variable), we will hold one variable (commonly $n$) fixed, and visualize one variable ($P$) as a function of all possible combinations of the others ($V,T$).

The plot below displays the $P(V,T)_{n=1.25}$ state space for for an ideal gas. The constant-volume slices you plotted in {ref}`2D-state` are indicated with lines. Get a feel for shape of the surface--you can rotate the figure and zoom in/out. 

````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio
from IPython.display import HTML


R = 0.08206  # L atm / mol K
n_1 = 1.25   # mol

T_values = np.linspace(79, 298, 80)
V_values = np.linspace(0.01, 1.0, 80)
curve_volumes = [1.0, 0.1, 0.01]

def ideal_pressure(T, V, n):
    return n * R * T / V


T_grid, V_grid = np.meshgrid(T_values, V_values)
P_grid1 = ideal_pressure(T_grid, V_grid, n_1)

# makes 3D subplots that can be interactive in-browser
fig = make_subplots(
    rows=1, cols=1,
    specs=[[{'type': 'surface'}]]
)

fig.add_trace(
    go.Surface(x=T_values, y=V_values, z=P_grid1, colorscale='Viridis', showscale=False, opacity=0.7),
    row=1, col=1
)


colors = ['blue', 'orange', 'red']
for V_val, color in zip(curve_volumes, colors):
    P_c1 = ideal_pressure(T_values, V_val, n_1)
    
    fig.add_trace(
        go.Scatter3d(x=T_values, y=np.full_like(T_values, V_val), z=P_c1,
                     mode='lines', line=dict(color=color, width=6), name=f'V = {V_val:g} L'),
        row=1, col=1
    )


scene_axis = dict(xaxis_title='Temperature (K)', yaxis_title='Volume (L)', zaxis_title='Pressure (atm)')
fig.update_layout(
    title_text='Ideal-gas P(V,T) State Surfaces',
    scene1=scene_axis,
    height=550,
    margin=dict(l=10, r=10, b=10, t=60)
)

#instead of fig.show(), to avoid breaking the $ math formatting
HTML(fig.to_html(
    include_plotlyjs='cdn',
    include_mathjax=False,
    full_html=False
))
````

## The van der Waals Equation of State

Johannes van der Waals derived a modified ideal gas law in an attempt to better capture the experimentally-observed behavior of gases. 

The assumption that the volume occupied by particles is neglibly different from the container volume ($V_\text{free} = V_\text{container} = V$) would only be true if the particles themselves take up no space.[^point-particle] However, in actuality gas particles _do_ take up space. van der Waals' equation of state aims to account for this by instead employing the **effective volume** ($V_\text{eff})$:

$$V_\text{eff} = V - nb,$$

where $b$ has units of volume per mole, representing the volume excluded by 1 mol of gas particles. By using $\bar V_{\text{eff}}$ in place of $V$, the equation of state becomes:

$$P=\frac{nRT}{V-nb}.$$

Several assumptions made in the derivation of the ideal gas law imply that gas particles do not attractively interact with one another. However, in actuality, particles can experience attractive interactions--and if they do, it will reduce their kinetic energy and thus the force with which they collide with container walls. The van der Waals' equation aims to correct for that by introducing a second parameter, $a$, which corrects the pressure based on the particle number density:

$$P_{\text{eff}}=P+ a\cdot \frac{n^2}{V^2}.$$

Substituting this in yields: 

$$P+\frac{a n^2}{V^2}=\frac{nRT}{V-nb},$$

or, equivalently,

$$P=\frac{nRT}{V-nb}-\frac{an^2}{V^2}.$$

This is known as the **van der Waals equation of state**. The $a$ and $b$ parameters are substance-specific and must be experimentally measured. Values for some substances can be found in table {numref}`vdw-params.

:class: hint margin simple

Be attentive to units! Recall:

````python
#cell metadata tagged: remove-cell

#PART A

import numpy as np
import matplotlib.pyplot as plt

Temps = np.array([323.15, 333.15, 343.15, 353.15, 363.15, 373.15]) #K
V1 = 0.50 #L
V2 = 0.30 #L 
V3 = 0.10 #L 

R = 0.08206 # L atm / mol K
a = 3.6 # L^2 bar / mol^2
b = 0.04267 # L / mol
n = 1 #mol

#convert units of a to match R: 
a = a * (0.9869/1) #bar * atm / bar = atm 

P_vdw_1 = n * R * Temps / (V1 - n*b) - a*n**2 / V1**2
P_vdw_2 = n* R * Temps / (V2 - n*b) - a*n**2 / V2**2
P_vdw_3 = n* R * Temps / (V3 - n*b) - a*n**2 / V3**2

figa = plt.figure(figsize = (4, 4))
plt.plot(Temps, P_vdw_1, 'o-', label = 'V = 0.50 L')
plt.plot(Temps, P_vdw_2, 'o-', label = 'V = 0.30 L')
plt.plot(Temps, P_vdw_3, 'o-', label = 'V = 0.10 L')
plt.xlabel('Temperature (K)')
plt.ylabel('Pressure (atm)')
plt.legend()

glue('co2-vdw-a', figa, display=False)
plt.close(figa)

#PART B

#all experimental pressures reported in atm
P_050 = [45, 45, 48, 50, 50, 53] #exp measured P at V = 0.50 L
P_030 = [65, 65, 70, 75, 80, 80] #exp measured P at V = 0.30 L
P_010 = [100, 125, 150, 150, 175, 175] #exp measured P at V = 0.10 L

diff_1 = 100 * P_vdw_1 / P_050
diff_2 = 100 * P_vdw_2 / P_030
diff_3 = 100 * P_vdw_3 / P_010

figb = plt.figure(figsize = (4, 4))
plt.plot(Temps, diff_1, 'o-', label = 'V = 0.50 L')
plt.plot(Temps, diff_2, 'o-', label = 'V = 0.30 L')
plt.plot(Temps, diff_3, 'o-', label = 'V = 0.10 L')
plt.xlabel('Temperature (K)')
plt.ylabel('% Difference (vdW - experimental) (atm)')
plt.legend()

glue('co2-vdw-b', figa, display=False)
plt.close(figb)
````

## Glossary 

:::{glossary}
mechanical properties
  Mechanical properties describe a physical object's relationships between force and motion. For a gas, key mechanical properties include pressure ($P$), volume ($V$), and number of moles ($n$). Notably, temperature is not a mechanical property (although it does influence a system's mechanical properties). 

equation of state
  A mathematical function that describes the state of a system, based on system parameters. 
  
:::
