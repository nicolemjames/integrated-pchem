# Wave Mechanics

Note: To use the interactives in this page, be sure to click on the rocket symbol at the top of the page, and select "live code".

For a statistical mechanics description of atomic and molecular energy states, we need a theory for describing the motion of atomic and molecular systems. One could use classical mechanics descriptions, essentially assuming that these macroscopic relationships will hold at the atomic and molecular scale. However, one could also take note that {term}`particle-wave duality` posits that matter (particularly atomic-scale matter) exibits wavelike motion. 

Notably, classical physics already has established theoretical models for the motion of waves (i.e., wave mechanics). If atomic-scale particules also behave as waves--and if classical wave mechanics is "true"--it should be possible to describe the motion of atomic-scale systems by applying classical wave mechanics.


(class-wave-1D)=
## The 1-D Classical Wave Equation

Let us consider a model system: an oscillating string fixed at two points, $x=0$ and $x=l$. Our objective is to determine a function that describes the motion of the string at any position and time: $\Psi(x, t)$. 

:::{admonition} Side Note
:class: hint margin simple

A **differential** is an infinitesimal change in a value $dy$. 

A **derivative** is an instanteous rate of change in one variable relative to to another (e.g.: $\frac{dy}{dx}$). 

An equation that describes the form of the infinitesimal change in a value is known as a **differential equation**. 

**Ordinary** differentials and derivatives involve single-variable functions, and are usually indicated with $d$. For example: $\frac{d}{dx} y(x)$. 

**Partial** differentials and derivatives involve multi-variable functions, and are usually indicated with $\partial$. For example: $\frac{\partial}{\partial x} y(x,z)$.  
:::

If we take a theoretical (i.e., not empirical) perspective to deriving a form for $\Psi(x, t)$, we can consider what relationships we _do_ know between $x$ and $t$: 

```{math}
:label: v
v = \frac{dx}{dt}
```

We can think of this as a constraint on the possible solutions of $\Psi(x,t)$. We can reformulate $\Psi(x,t)$ in terms of this constraint:


```{math}
:label: second-deriv
\frac{\partial ^2}{\partial x ^2} \Psi(x,t) = \frac{\partial}{\partial x} \frac{\partial}{\partial x} \Psi(x,t)
```

Now, we can solve Equation {eq}`v` for  $\partial x$, and substitute in to Equation {eq}`second-deriv`:

```{math}
:label: d2adx2

\frac{\partial ^2 \Psi}{\partial x ^2} = \frac{1}{v^2} \frac{\partial ^2 \Psi}{\partial t^2}
```

This is now a slightly more specific constraint for what form a solution for $\Psi(x,t)$ could take. There are also two additional constraints we can determine, from our physical understanding of the system. Because our system describes the motion of a string fixed at $x=0$ and $x=l$, it must be the case that:


```{math}
:label: a(0,t)
\Psi(0,t) = 0 
```

```{math}
:label: a(l,t)
\Psi(l,t) = 0 
```
:::{admonition} Side Note
:class: hint margin simple

Equations {eq}`a(0,t)` and {eq}`a(l,t)` are {term}`boundary conditions`
:::

Thus, we now have three constraints on what form the equation of motion $\Psi(x,t)$ can take, given by Equations {eq}`d2adx2`, {eq}`a(0,t)`, and {eq}`a(l,t)`.

In order to find a function form of $\Psi(x,t)$ that satisfies these constraints, let us assume that the variables $x$ and $t$ are separable. In other words: we assume that $\Psi(x,t)$ is a product of two functions, each depending on only one variable: 

```{math}
:label: prod
\Psi(x,t) = f(x) g(t)
```

Substituting Equation {eq}`prod` into Equation {eq}`d2adx2` yields:

```{math}
:label: dx2fgdx2
\frac{\partial^2}{\partial x^2} f(x)g(t)= \frac{1}{v^2} \frac{\partial^2}{\partial t^2} f(x)g(t)
```

Because $f(x)$ depends only on $x$, and $g(t)$ depends only on $t$, we can collect variables on each side by dividing both sides by $\Psi(x,t) = f(x)g(t)$:

```{math}
:label: dx2-sep
\frac{1}{f(x)} \frac{\partial^2}{\partial x^2} f(x)= \frac{1}{v^2 g(t)} \frac{\partial^2}{\partial t^2} g(t)
```

Because the left hand side contains no $t$-dependence, and the right-hand side contains no $x$-dependence, they can each vary independently of the other. The only way for this to be true and both sides to still equal each other, is if they both equal the same constant ($K)$. This is called a "separation constant," because it can be used to separate Equation {eq}`dx2-sep` into two equations:

::::{grid} 1 1 2 2
:gutter: 3

:::{grid-item-card}
:class-card: sd-shadow-sm

```{math}
:label: dx2-sepd
\frac{1}{f(x)} \frac{\partial^2}{\partial x^2} f(x) = K 
```
:::

:::{grid-item-card} 
:class-card: sd-shadow-sm

```{math}
:label: dt2-sepd
\frac{1}{v^2 g(t)} \frac{\partial^2}{\partial t^2} g(t) = K 
```
:::
::::

Which can be re-expressed: 

::::{grid} 1 1 2 2
:gutter: 3

:::{grid-item-card}
:class-card: sd-shadow-sm

```{math}
:label: dx2-sepd-2
\frac{\partial^2}{\partial x^2} f(x) - f(x)K=0 
```
:::

:::{grid-item-card} 
:class-card: sd-shadow-sm

```{math}
:label: dt2-sepd-2
\frac{\partial^2}{\partial t^2} g(t) - v^2 g(t)K = 0 
```
:::
::::

Both expressions are of the form:

```{math}
:label: gen-form
\frac{\partial^2}{\partial x^2} y(x) - y(x) K = 0 
```
To solve this, we must separately consider three possibilities:

::::{grid} 1 1 3 3
:gutter: 3

:::{grid-item-card}
:class-card: sd-shadow-sm
$K > 0 $
:::

:::{grid-item-card}
:class-card: sd-shadow-sm
$K = 0 $
:::

:::{grid-item-card}
:class-card: sd-shadow-sm
$K < 0 $
:::
::::

The $K = 0$ case is trivial: if $K = 0$, then $\frac{\partial ^2}{\partial x^2} y(x)=0$. Let us consider the cases where $K \neq 0$. "Experience shows" that differential equations of the type shown in Equation {eq}`gen-form` have solutions of the form: 

$$y(x) = e^{\alpha x}$$

where $\alpha$ is a constant. Thus:  


::::{grid} 1 1 2 2
:gutter: 3

:::{grid-item-card} $K > 0$
:class-card: sd-shadow-sm

```{math}
\frac{\partial^2}{\partial x^2} y(x) - y(x) K = 0 
```

Substituting in the general solution (Equation {eq}`gen-form`):

```{math}
\frac{\partial^2}{\partial x^2} e^{\alpha x} - e^{\alpha x} \cdot K = 0
```

Computing the derivatives yields:

```{math}
\alpha^2 e^{\alpha x} - e^{\alpha x} \cdot K = 0
```

which can be simplified:

```{math}
(\alpha^2 - K)e^{\alpha x} = 0
```

and solved for $\alpha$:

$\alpha ^2 = K$
\
$\alpha = \pm \sqrt{K} = \pm k$ [^renam]

Yielding two possible solutions:

```{math}
y_1(x) = e^{kx} 

y_2(x) = e^{-kx} 
```

The general solution can be expressed as a linear combination of both solutions, weighted by constants $C_1$ and $C_2$:

```{math}
:label: K>0-soln 
y(x) = C_1 e^{kx} + C_2 e^{-kx} 
```
:::

:::{grid-item-card} $K < 0$
:class-card: sd-shadow-sm

```{math}
\frac{\partial^2}{\partial t^2} y(x) + y(x) K = 0 
```
Substituting in the general solution (Equation {eq}`gen-form`):

```{math}
\frac{\partial^2}{\partial x^2} e^{\alpha x} + e^{\alpha x} \cdot K = 0
```

Computing the derivatives yields:

```{math}
\alpha^2 e^{\alpha x} + e^{\alpha x} \cdot K = 0
```

which can be simplified:

```{math}
(\alpha^2 + K)e^{\alpha x} = 0 
```

and solved for $\alpha$:

$\alpha ^2 = -K$
\
$\alpha = \pm i \sqrt{K} = \pm i k$ [^renam]

Yielding two possible solutions:

```{math}
y_1(x) = e^{ikx} 

y_2(x) = e^{-ikx} 
```

The general solution can be expressed as a linear combination of both solutions, weighted by constants $C_3$ and $C_4$:

```{math}
:label: K<0-soln1
y(x) = C_3 e^{ikx} + C_4 e^{-ikx}
```

By using {term}`Euler's formula`, Equation {eq}`K<0-soln1` can be written:

```{math}
:label: K<0-soln2
y(x) = C_5 \cos (kx) + C_6 \sin(kx)
```
:::
::::

Thus, we have two possible sets of solutions to Equations {eq}`dx2-sepd-2` and {eq}`dt2-sepd-2`; to narrow our solution set further, we invoke our boundary conditions. 

Starting with {eq}`dx2-sepd-2`:  

::::{grid} 1 1 2 2
:gutter: 3

:::{grid-item-card} $K > 0$
:class-card: sd-shadow-sm

```{math}
f(x) = A e^{\beta x} + B e^{\beta x}
```
Via our boundary conditions, we know $f(0) = 0$ and $f(l) = 0$:

```{math}
f(0) = A e^{0} + B e^{0} = 0 
```
Because $e^0 = 1$, the only way this can be true is if $A=B=0$. Thus, this is a second trivial solution. 
:::

:::{grid-item-card} $K < 0$
:class-card: sd-shadow-sm

```{math}
f(x) = A \cos (\beta x) + B \sin (\beta x)
```
Via our boundary conditions, we know $f(0) = 0$ and $f(l) = 0$:

```{math}
f(0) = A \cos (0) + B \sin (0)=0
```

```{math}
f(0) = A + 0 = 0 
```
Thus, $A$ must $=0$, and our expression simplifies to:

```{math}
f(x) = B \sin (\beta x)
```

We can use our second boundary condition, $f(l)=0$ to determine a functional form of $\beta$:

```{math}
f(l) = B \sin (\beta l)=0
```

$\sin(x)=$ when $x=n\pi$, where $n$ is an integer. Thus, $\beta l$ must $=n\pi$:

```{math}
:label: beta
\beta = \frac{n\pi}{l}
```

yielding:

```{math}
:label: soln-1
f(x) = B \sin \left(\frac{n\pi}{l} x\right)
```
:::
::::

Now we repeat this process for Equation {eq}`dt2-sepd-2`. Since $K$ must be the same for both $f(x)$ and $g(t)$, we can express Equation {eq}`dt2-sepd-2` as:

```{math}
:label: dt2-sepd-3
\frac{\partial^2}{\partial t^2} g(t) - \beta^2 v^2 g(t) = 0 
```

And because we know that non-trivial solutions require $K < 0$, our solution must be of the form: 

```{math}
:label: dt2-sepd-soln
g(t) = D \cos (\beta v t) + E \sin (\beta v t)
```

We do not have boundary conditions for $t$, so we cannot resolve expressions for $D$ and $E$, but they may depend on $n$. The term $\beta v$ is the {term}`angular frequency`, commonly represented as $\omega$:

$$\omega = \frac{n \pi v}{l}$$

We can simplify this further using the relation:

$$\cos(ax+c) = \cos(ax)\cos(c) - \sin(ax)\sin(c)$$

Applying this to Equation {eq}`dt2-sepd-soln`, it becomes:

```{math}
:label: soln-2
g(t) = A_n \cos (\omega_n t + \phi_n)
```
$A$ and $\phi$ are constants that can be expressed in terms of $D$ and $E$. Because $D$ and $E$ may depend on $n$, $A$ and $\phi$ may also depend on $n$. To remember this, we note them $A_n$ and $\phi_n$. 

Using Equations {eq}`soln-1`, {eq}`soln-2`, and {eq}`prod`, we can now have a solution for $\Psi(x,t)$: 

```{math}
:label: a(x,t)
\Psi(x,t) = A_n \cos\left(\omega_n t + \phi_n\right) \cdot \sin \left( \frac{n\pi}{l}x \right)
```

Note, however, this solution is still actually an infinite number of solutions, one for each value of $n$. To formulate a general solution, we can take a linear combination of all solutions, wrapping our weighting constant into $A_n$: 

$$ \Psi(x,t) = A_1\cos\left(\omega_1 t+ \phi_1 \right) \cdot \sin \left( \frac{\pi}{l} x \right) + A_2\cos\left(\omega_2 t+ \phi_2 \right) \cdot \sin \left( \frac{2 \pi}{l} x \right) + A_3\cos\left(\omega_3 t+ \phi_3 \right) \cdot \sin \left( \frac{3\pi}{l} x \right) + ... $$

Each term represents a "normal mode" of motion. Any motion can be represented as a linear combination of normal modes:

```{math} 
:label: 1d-wave
:class: highlighted-equation
\Psi(x,t) = \sum_{n=1}^{\infty} A_n\cos\left(\omega_n t+ \phi_n \right) \cdot \sin \left( \frac{n\pi}{l} x \right)
```

where each $n$ defines each normal mode, and time dependence is modified by the {term}`angular frequency`:

$$\omega_n = \beta v = \frac{n \pi v}{l}$$

[^renam]: We simply renamed the constant $\sqrt{K}$ to avoid having to write out square roots. 

Let's see what each of the first few normal modes look like. Expand the code cell below, scroll to the bottom, and click "run". Make sure you've activated live coding first! 

````python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

plt.ioff()
Lx, X = 1.0, np.linspace(0, 1.0, 40)

fig, ax = plt.subplots(figsize=(5, 5))
for attr in ['header_visible', 'footer_visible']:
    if hasattr(fig.canvas, attr): setattr(fig.canvas, attr, False)

ax.set(xlim=(0, Lx), ylim=(-1.05, 1.05), xlabel="x", ylabel="Ψ")
plt.tight_layout()
line, = ax.plot(X, np.zeros_like(X), color='black')

def draw_curve(n, frame):    
    line.set_ydata(np.sin(n * np.pi * X / Lx) * np.cos(np.pi * frame / 50))
    fig.canvas.draw_idle()

n_box = widgets.BoundedIntText(value=1, min=1, max=10, description="n:")
f_slider = widgets.BoundedIntText(value=0, min=0, max=100, description="frame:")
play = widgets.Play(value=0, min=0, max=100, interval=50)
widgets.jslink((play, "value"), (f_slider, "value"))

out = widgets.interactive_output(draw_curve, {"n": n_box, "frame": f_slider})
ctrls = widgets.VBox([n_box, f_slider, play], layout=widgets.Layout(align_items='center'))

display(widgets.VBox(
    [ctrls, out, fig.canvas], 
    layout=widgets.Layout(align_items='center', width='max-content')
))
````

Below, let's explore the types of motion we can create from the linear combination of three modes ($a$, $b$, $c$). 

Note that we have normalized the amplitude of the wave, so there is a constraint: $\sqrt{A_a^2 + A_b^2 + A_c^2} = 1$. If you adjust one amplitude, the others will be adjusted relative to this constraint, in the way that best retains their relative values. 

````python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

plt.ioff()
Lx, X = 1.0, np.linspace(0, 1.0, 40)

fig, ax = plt.subplots(figsize=(5, 5))
for attr in ['header_visible', 'footer_visible']:
    if hasattr(fig.canvas, attr): setattr(fig.canvas, attr, False)

ax.set(xlim=(0, Lx), ylim=(-1.75, 1.75), xlabel="x", ylabel="Ψ")
plt.tight_layout()
line, = ax.plot(X, np.zeros_like(X), color='black')

def draw_curve(na, nb, nc, a, b, c, frame):    
    phase = 2 * np.pi * frame / 100
    Y = (a*np.sin(na*np.pi*X/Lx) + b*np.sin(nb*np.pi*X/Lx) + c*np.sin(nc*np.pi*X/Lx)) * np.cos(phase)
    line.set_ydata(Y)
    fig.canvas.draw_idle()

bx_lyt = widgets.Layout(width='150px')
n_args = dict(min=1, max=10, layout=bx_lyt)
na_box = widgets.BoundedIntText(value=1, description="n_a:", **n_args)
nb_box = widgets.BoundedIntText(value=2, description="n_b:", **n_args)
nc_box = widgets.BoundedIntText(value=3, description="n_c:", **n_args)

A_args = dict(min=0.0, max=1.0, step=0.01, readout_format='.2f', continuous_update=False)
Aa_sl = widgets.FloatSlider(value=1.0, description="A_a:", **A_args)
Ab_sl = widgets.FloatSlider(value=0.0, description="A_b:", **A_args)
Ac_sl = widgets.FloatSlider(value=0.0, description="A_c:", **A_args)

f_box = widgets.BoundedIntText(value=0, min=0, max=100, description="phase:", layout=bx_lyt)
play = widgets.Play(value=0, min=0, max=100, interval=50)
widgets.jslink((play, "value"), (f_box, "value"))

def normalize_weights(change):
    sliders = (Aa_sl, Ab_sl, Ac_sl)
    for s in sliders: s.unobserve(normalize_weights, names='value')
    
    new_val = max(0.0, min(1.0, change['new']))
    if change['owner'] == Aa_sl: s1, s2 = Ab_sl, Ac_sl
    elif change['owner'] == Ab_sl: s1, s2 = Aa_sl, Ac_sl
    else: s1, s2 = Aa_sl, Ab_sl
        
    v1, v2 = s1.value, s2.value
    t_sq = max(0.0, 1.0 - new_val**2)
    
    if t_sq <= 1e-5:
        s1.value = s2.value = 0.0
    else:
        c_sq = v1**2 + v2**2
        if c_sq > 1e-5:
            scale = np.sqrt(t_sq / c_sq)
            s1.value, s2.value = round(v1 * scale, 2), round(v2 * scale, 2)
        else:
            s1.value = s2.value = round(np.sqrt(t_sq / 2.0), 2)

    for s in sliders: s.observe(normalize_weights, names='value')

for s in (Aa_sl, Ab_sl, Ac_sl): s.observe(normalize_weights, names='value')

out = widgets.interactive_output(draw_curve, {
    "na": na_box, "nb": nb_box, "nc": nc_box, 
    "a": Aa_sl, "b": Ab_sl, "c": Ac_sl, "frame": f_box
})

ctrl_panel = widgets.VBox([
    widgets.HTML("<b>Adjust Modes and Amplitudes:</b>"),
    widgets.HBox([na_box, Aa_sl]), 
    widgets.HBox([nb_box, Ab_sl]), 
    widgets.HBox([nc_box, Ac_sl]),
    widgets.HBox([play, f_box], layout=widgets.Layout(justify_content='center', margin='10px 0'))
], layout=widgets.Layout(align_items='center'))

display(widgets.VBox([ctrl_panel, out, fig.canvas], layout=widgets.Layout(align_items='center', width='max-content')))
````

## The 2-D Classical Wave Equation

Following the same process as in {numref}`class-wave-1D`, we can describe the motion of a wave on a 2-D rectangular membrane by solving the partial differential equation:

```{math}
\frac{\partial^2}{\partial x^2} \Psi(x,y,t) + \frac{\partial^2}{\partial y^2} \Psi(x,y,t) = \frac{1}{v^2} \frac{\partial^2}{\partial t^2} \Psi(x,y,t)
```

Which can be solved by an equation of the form: 

```{math} 
:label: 2d-wave
:class: highlighted-equation
\Psi(x,y,t) = \left[ A_{n_x,n_y} \cos \left(\omega_{n_x, n_y} t + \phi_{n_x, n_y} \right) + B_{n_x,n_y} \sin \left( \omega_{n_x,n_y} t+ \phi_{n_x, n_y} \right) \right]  \cdot \sin \left( \frac{n_x \pi}{l_x} x \right) \cdot \sin \left( \frac{n_y \pi}{l_y} y \right)
```

Thus, the general solution is: 

```{math} 
:label: 2d-wave-gen
:class: highlighted-equation
\Psi(x,y,t) = \sum_{n_x=1}^{\infty}  \sum_{n_y=1}^{\infty} \left[ A_{n_x,n_y} \cos \left(\omega_{n_x, n_y} t + \phi_{n_x, n_y} \right) + B_{n_x,n_y} \sin \left( \omega_{n_x,n_y} t+ \phi_{n_x, n_y} \right) \right]  \cdot \sin \left( \frac{n_x \pi}{l_x} x \right) \cdot \sin \left( \frac{n_y \pi}{l_y} y \right)
```

Note the derivation of these expressions still involves assuming separability of the spatial and time components. Thus, we can also present the "time-independent" (spatial-only) solution: 

```{math} 
:label: 2d-wave-final
:class: highlighted-equation
\Psi(x,y) =  \sin \left( \frac{n_x \pi}{l_x} x \right) \cdot \sin \left( \frac{n_y \pi}{l_y} y \right)
```

Or, as the general solution:

```{math} 
:label: 2d-wave-gen-final
:class: highlighted-equation
\Psi(x,y) = \sum_{n_x=1}^{\infty}  \sum_{n_y=1}^{\infty} \sin \left( \frac{n_x \pi}{l_x} x \right) \cdot \sin \left( \frac{n_y \pi}{l_y} y \right)
```

Let's examine. Again, make sure live coding is active, expand the code cell below, and click "run". 

````python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
from matplotlib.colors import LinearSegmentedColormap, Normalize

plt.ioff()
Lx, Ly = 1.0, 1.0
X, Y = np.meshgrid(np.linspace(0, Lx, 35), np.linspace(0, Ly, 35))
cmap, norm = LinearSegmentedColormap.from_list("m", [(0,0,1,1), (1,1,1,0), (1,0,0,1)]), Normalize(-1, 1)


fig = plt.figure(figsize=(5, 5))
ax = fig.add_subplot(111, projection="3d")
for attr in ['header_visible', 'footer_visible']:
    if hasattr(fig.canvas, attr): setattr(fig.canvas, attr, False)

ax.set(xlim=(0, Lx), ylim=(0, Ly), zlim=(-1.05, 1.05), xlabel="x", ylabel="y", zlabel="Ψ")
plt.tight_layout()

def draw_membrane(nx, ny, frame):
    [c.remove() for c in list(ax.collections)]
    Z = (np.sin(nx * np.pi * X / Lx) * np.sin(ny * np.pi * Y / Ly)) * np.cos(2 * np.pi * frame / 100)
    ax.plot_surface(X, Y, Z, facecolors=cmap(norm(Z)), rstride=1, cstride=1, edgecolor='black', linewidth=0.5, antialiased=True, shade=True)
    fig.canvas.draw_idle()

bx_lyt = widgets.Layout(width='150px')
nx_box = widgets.BoundedIntText(value=1, min=1, max=10, description="n_x", layout=bx_lyt)
ny_box = widgets.BoundedIntText(value=1, min=1, max=10, description="n_y", layout=bx_lyt)
f_box = widgets.BoundedIntText(value=0, min=0, max=100, description="frame", layout=bx_lyt)

play = widgets.Play(value=0, min=0, max=100, interval=50)
widgets.jslink((play, "value"), (f_box, "value"))

out = widgets.interactive_output(draw_membrane, {"nx": nx_box, "ny": ny_box, "frame": f_box})
ctrl_panel = widgets.VBox([nx_box, ny_box, f_box, play], layout=widgets.Layout(align_items='center'))

display(widgets.VBox([ctrl_panel, out, fig.canvas], layout=widgets.Layout(align_items='center', width='max-content')))
````

As before, let's explore the types of motion we can create from the linear combination of three modes ($a$, $b$, $c$):

````python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
from matplotlib.colors import LinearSegmentedColormap, Normalize

plt.ioff()
Lx, Ly = 1.0, 1.0
X, Y = np.meshgrid(np.linspace(0, Lx, 35), np.linspace(0, Ly, 35))
cmap, norm = LinearSegmentedColormap.from_list("m", [(0,0,1,1), (1,1,1,0), (1,0,0,1)]), Normalize(-1.5, 1.5)

fig = plt.figure(figsize=(5, 5))
ax = fig.add_subplot(111, projection="3d")
for attr in ['header_visible', 'footer_visible']:
    if hasattr(fig.canvas, attr): setattr(fig.canvas, attr, False)

ax.set(xlim=(0, Lx), ylim=(0, Ly), zlim=(-1.05, 1.05), xlabel="x", ylabel="y", zlabel="Ψ")
plt.tight_layout()

def draw_membrane(nxa, nya, nxb, nyb, nxc, nyc, a, b, c, frame):
    [c.remove() for c in list(ax.collections)]
    mode_a = a * np.sin(nxa * np.pi * X / Lx) * np.sin(nya * np.pi * Y / Ly)
    mode_b = b * np.sin(nxb * np.pi * X / Lx) * np.sin(nyb * np.pi * Y / Ly)
    mode_c = c * np.sin(nxc * np.pi * X / Lx) * np.sin(nyc * np.pi * Y / Ly)
    Z = (mode_a + mode_b + mode_c) * np.cos(2 * np.pi * frame / 100)
    
    ax.plot_surface(X, Y, Z, facecolors=cmap(norm(Z)), rstride=1, cstride=1, edgecolor='black', linewidth=0.3, antialiased=True, shade=True)
    fig.canvas.draw_idle()

bx_lyt = widgets.Layout(width='120px')
b_kw = dict(min=1, max=10, layout=bx_lyt)
nxa_b, nya_b = widgets.BoundedIntText(value=1, description="nxa:", **b_kw), widgets.BoundedIntText(value=1, description="nya:", **b_kw)
nxb_b, nyb_b = widgets.BoundedIntText(value=2, description="nxb:", **b_kw), widgets.BoundedIntText(value=2, description="nyb:", **b_kw)
nxc_b, nyc_b = widgets.BoundedIntText(value=3, description="nxc:", **b_kw), widgets.BoundedIntText(value=3, description="nyc:", **b_kw)

s_kw = dict(min=0.0, max=1.0, step=0.01, readout_format='.2f', continuous_update=False)
Aa_sl, Ab_sl, Ac_sl = widgets.FloatSlider(value=1.0, description="Aa:", **s_kw), widgets.FloatSlider(value=0.0, description="Ab:", **s_kw), widgets.FloatSlider(value=0.0, description="Ac:", **s_kw)

f_box = widgets.BoundedIntText(value=0, min=0, max=100, description="phase:", layout=bx_lyt)
play = widgets.Play(value=0, min=0, max=100, interval=50)
widgets.jslink((play, "value"), (f_box, "value"))

sliders = (Aa_sl, Ab_sl, Ac_sl)
def normalize_weights(change):
    for s in sliders: s.unobserve(normalize_weights, names='value')
    new_val = max(0.0, min(1.0, change['new']))
    
    if change['owner'] == Aa_sl: s1, s2 = Ab_sl, Ac_sl
    elif change['owner'] == Ab_sl: s1, s2 = Aa_sl, Ac_sl
    else: s1, s2 = Aa_sl, Ab_sl
        
    v1, v2 = s1.value, s2.value
    t_sq = max(0.0, 1.0 - new_val**2)
    
    if t_sq <= 1e-5:
        s1.value = s2.value = 0.0
    else:
        c_sq = v1**2 + v2**2
        if c_sq > 1e-5:
            scale = np.sqrt(t_sq / c_sq)
            s1.value, s2.value = round(v1 * scale, 2), round(v2 * scale, 2)
        else:
            s1.value = s2.value = round(np.sqrt(t_sq / 2.0), 2)

    for s in sliders: s.observe(normalize_weights, names='value')

for s in sliders: s.observe(normalize_weights, names='value')

out = widgets.interactive_output(draw_membrane, {
    "nxa": nxa_b, "nya": nya_b, "nxb": nxb_b, "nyb": nyb_b, "nxc": nxc_b, "nyc": nyc_b,
    "a": Aa_sl, "b": Ab_sl, "c": Ac_sl, "frame": f_box
})

ctrl_panel = widgets.VBox([
    widgets.HTML("<b>Adjust Modes and Amplitudes:</b>"),
    widgets.HBox([nxa_b, nya_b, Aa_sl]), 
    widgets.HBox([nxb_b, nyb_b, Ab_sl]), 
    widgets.HBox([nxc_b, nyc_b, Ac_sl]),
    widgets.HBox([play, f_box], layout=widgets.Layout(justify_content='center', margin='10px 0'))
], layout=widgets.Layout(align_items='center'))

display(widgets.VBox([ctrl_panel, out, fig.canvas], layout=widgets.Layout(align_items='center', width='max-content')))
````

## Glossary 

:::{glossary}
particle-wave duality
  Light and matter exhibit both particle-like and wave-like behaviors. The particle-like nature of light (electromagnetic radiation) was demonstrated through the photoelectric effect. Louis de Broglie theorized that it could then be true that particulate matter might also exhibit wave-like behaviors, and derived a prediction for the wavelength of a particle ($\lambda = h/\rho$, where $\lambda$ is the wavelength, $h$ is Planck's constant, and $\rho$ is the particle momentum.) This prediction shows that the wavelike nature of matter is only relevant for very small particles, but this bears considerable implications for the motion of atomic-scale particulates such as electrons. 

partial differential equation 
  An equation involving the partial derivatives of an unknown function. An expression for the function that satisfies the partial differential equation (PDE) is considered a solution to the PDE. There may be multiple solutions to a PDE, analogous to how the equation $9 = \sqrt{x^2}$ has two solutions ($x = +3, -3$)

boundary conditions
  known values or relationships that apply the edges (boundaries) of asystem 

angular frequency
  rate of change in the value of an angle (angle per unit time)

Euler's formula
  $e^{ix} = \cos(x) + i \sin(x)$
